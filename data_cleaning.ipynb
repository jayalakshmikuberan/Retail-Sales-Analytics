```python
import pandas as pd

# Load raw dataset
df = pd.read_csv("retail_sales_dataset.csv")

print("--- ORIGINAL DATASET SHAPE ---")
print(df.shape)
```

```python
# Check and drop missing values
missing_initial = df.isnull().sum().sum()
print(f"Missing values found: {missing_initial}")
df = df.dropna()

# Check and remove duplicates
duplicates_initial = df.duplicated().sum()
print(f"Duplicate rows found: {duplicates_initial}")
df = df.drop_duplicates()
```

```python
# Standardize column names to lowercase with underscores
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')

print("--- STANDARDIZED COLUMNS ---")
print(df.columns.tolist())

# Convert date column to standard datetime format
df['date'] = pd.to_datetime(df['date'], format='%Y-%m-%d')

print("\n--- DATE RANGE ---")
print("Min Date:", df['date'].min())
print("Max Date:", df['date'].max())
```

```python
# Extract date components for deeper trend analysis
df['year'] = df['date'].dt.year
df['month'] = df['date'].dt.month
df['day'] = df['date'].dt.day
df['day_of_week'] = df['date'].dt.day_name()

# Recalculate total amount to ensure data integrity
df['total_amount'] = df['quantity'] * df['price_per_unit']

print("--- FINAL DATA TYPES ---")
print(df.dtypes)
```

```python
print("--- FINAL DATASET PREVIEW ---")
print(df.head())

print("\n--- SUMMARY STATISTICS ---")
print(df.describe())

# Save final cleaned dataset for the SQL team member
df.to_csv("cleaned_sales.csv", index=False)
print("\nSuccess: Final cleaned dataset saved as 'cleaned_sales.csv'!")

# Final Cleaning Report
print("\n--- CLEANING REPORT ---")
print(f"Original rows:    1000") # Adjust if your starting row count is different
print(f"Final rows:       {df.shape[0]}")
print(f"Rows removed:     {1000 - df.shape[0]}")
print(f"Final columns:    {df.shape[1]}")
```
